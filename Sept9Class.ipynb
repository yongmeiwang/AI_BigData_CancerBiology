```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y=x**2
fig, ax = plt.subplots()
ax.plot (x,y)
ax.set_xlabel("this is my x")
ax.set_ylabel("y axis ")
plt.show()
```

```python
# Define the two functions
y1 = x**2
y2 = x**3
x=np.linspace(-2,2)
# Plot both functions
plt.plot(x, y1, label='y1 = x²')
plt.plot(x, y2, label='y2 = x³')

# Add labels and legend
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of y = x² and y = x³')
plt.legend()
plt.grid(True)

plt.show()

```

Now we calculate length of each chromosomes. We define one basepair length, bp_length as 3.4 Agnstrom. then calculate the legnth of each chromosome and store in the third column.

```python
# Length of one base pair in Angstroms
bp_length = 3.4

# Calculate chromosome length
chrom_data["Length (Angstrom)"] = chrom_data["basepais(bp)"] * bp_length

# Display the result
chrom_data.head()
```

```python
total_length_angstrom = chrom_data["Length (Angstrom)"].sum()
total_length_meter = total_length_angstrom / 1e10
print(f"Total length of all chromosomes in meters: {total_length_meter} m")
```
